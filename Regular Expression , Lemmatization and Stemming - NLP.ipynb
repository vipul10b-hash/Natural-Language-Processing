```python
# Remove websites links and URLs - \S matches all non-space characters and characters like :// in the URL removed
url_text = "This is one type of url: https://www.wikipedia.com and the other url is http://www.wikipedia.com"
print(re.findall("(http\S+)",url_text))
```

```python
#Remove email addresses - Email addresses can be written in diffrent forms such as name - surname@domain.(suffix),
#name_surname@domain.(suffix) or name-surname@domain.(suffix) etc.,

email_text = "Please contact us at priyanka-madiraju7445@company.com or priyanka.madiraju@company.com or on the phone number"
print(re.findall("([\w\.\-\_]+@[\w\.\-\_]+)",email_text))
```

```python
# Remove digits
numbers_text = "Phone Numbers are 33635353 and 9383777"
print(re.sub("(\d+)","",numbers_text))
```

```python
#Remove newline characters - sometimes the text has formatting like the new line \n and these need to be removed.
newline_text="And I in going ,madam,weep o'er my father's death\n\n anew: I am now in ward,evermore in subjection."
print(re.sub("(\n+)","",newline_text))
```

```python
#Remove multiple spaces within text
url_text_output = "This is one type of url: and the other url is"
print(re.sub("(\s){2,}"," ",url_text_output))
```

```python
#Remove special characters except for space
url_text_output = "This is one type of url: # ? and the other url is"
print(re.sub("([^a-zA-Z\s])","",url_text_output))
```

## Word Lemmatization
- As a defination , lemmatization is to "remove inflectional endings only and to returen the base or dictionary form of word , which is known as the lemma."
- First,we create a raw text and tokenize it into tokens using the function word_tokenize.

```python
import nltk
```

```python
nltk.download('wordnet')
```

```python
import nltk
nltk.download('punkt_tab')
wn1 = nltk.WordNetLemmatizer()
text = "studies studying cries cry"
tokenization = nltk.word_tokenize(text)
for w in tokenization:
    print("Lemma for {} is {}".format(w,wn1.lemmatize(w)))
```

## Stemming

```python
from nltk.stem.porter import PorterStemmer  #---> Pronunciation will same but may not have meaningfull word
ps = PorterStemmer()
text = "studies studying cries cry"
tokenization = nltk.word_tokenize(text)
for w in tokenization:
    print("Stemming for {} is {}".format(w,ps.stem(w)))
```

```python
from nltk.stem.lancaster import LancasterStemmer  #--> Pronunciation will same but have meaningfull word
lc = nltk.LancasterStemmer()
text = "studies studying cries cry"
tokenization = nltk.word_tokenize(text)
for w in tokenization:
    print("Stemming for {} is {}".format(w,lc.stem(w)))
```
