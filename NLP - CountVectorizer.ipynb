# Natural Language Processing (NLP)

## CountVectorizer 
--> To convert text into number

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
text = ['Hello my name is james','james this is my python notebook','james trying to create a big dataset','james of words to try differnt','features of count vectorizer']
```

```python
count_vect = CountVectorizer()
```

```python
count_matrix = count_vect.fit_transform(text)
```

```python
count_array = count_matrix.toarray()
```

```python
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
```

```python
df
```

## Parameter
- Lowercase

--> Convert all characters to lowercase before tokenizing . Default is set to true and takes boolean value.

```python
text = ['hello my name is james','Hello my name is James']
```

```python
count_vect = CountVectorizer(lowercase=False)
```

```python
count_matrix = count_vect.fit_transform(text)
```

```python
count_array = count_matrix.toarray()
```

```python
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
```

```python
df
```

- Stop_words

```python
text = ['Hello my name is james','james this is my python notebook','james trying to create a big dataset','james of words to try differnt','features of count vectorizer']
```

```python
count_vect = CountVectorizer(stop_words= ['is','to','my'])
count_matrix = count_vect.fit_transform(text)
```

```python
count_array = count_matrix.toarray()
```

```python
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
```

```python
df
```

- stop_words = Whole english language (helping verb like 370 words)

```python
count_vect = CountVectorizer(stop_words= 'english')
count_matrix = count_vect.fit_transform(text)
count_array = count_matrix.toarray()
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
df
```

## Using max_df and min_df(covered later)


- Max_df:

```python
count_vect = CountVectorizer(max_df=1)
count_matrix = count_vect.fit_transform(text)
count_array = count_matrix.toarray()
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
df
```

```python
count_vect = CountVectorizer(max_df=0.75)
count_matrix = count_vect.fit_transform(text)
count_array = count_matrix.toarray()
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
df
# Here 0.75 is 75% --> occuring the word 
```

- Min_df:

```python
count_vect = CountVectorizer(min_df =2)
count_matrix = count_vect.fit_transform(text)
count_array = count_matrix.toarray()
df = pd.DataFrame(data=count_array,columns = count_vect.get_feature_names_out())
df
```
